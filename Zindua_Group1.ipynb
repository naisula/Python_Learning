APIS 
Application Programming Interface - a way for different software systems to communicate with each other.
Example:
Uber uses Google Maps → API sends location data

The “Contract” Idea
An API works like a contract between two systems:
Provider (Server): Promises to give data if requested correctly
Consumer (Client): Must request data in the correct format
If the client sends the wrong request (e.g., invalid city name), the server rejects it.
If the request is correct, the server responds with data—without revealing how it works internally.

How APIs Work
1. Request (Client → Server)
The client asks for data using:
URL (Endpoint): Where to go (e.g. /users)
Method:
GET → retrieve data
POST → send data
Headers: Extra info (like authentication)
Body: Data being sent (mainly for POST/PUT)

2. Processing (Server side)
The server:

Checks if request is valid
Fetches data from database
Formats the result
3.Response (Server → Client)

The server sends back:

Status Code: Shows success or error (e.g., 200 = OK, 404 = Not found)
Headers: Extra info (like rate limits)
Body: The actual data (often in JSON format)

Think of an API like ordering food:

You = Client
Restaurant = Server
Menu rules = API contract
Order = Request
Food delivered = Response

You don’t need to know how the kitchen works—just how to order correctly.

The Data Format: JSON
While APIs can return XML or HTML, 95% of modern APIs return JSON (JavaScript Object Notation).
Why JSON? It is lightweight and text-based.
The Python Connection: JSON maps perfectly to Python data structures.
JSON Object {}  →  → Python Dictionary dict
JSON Array []   →  → Python List list
JSON null       →  → Python None
JSON true       →  → Python True

HTTP Status Codes: The Server's Language
You must memorize the categories to debug effectively.
200 OK: The standard success response.
201 Created: Success, specifically for a POST request (a new record was made).
204 No Content: Success, but there is nothing to return (common for DELETE).
400 Bad Request: You sent invalid JSON or missing parameters.
401 Unauthorized: You are missing an API Key or Password.
403 Forbidden: You have a key, but you don't have permission for this specific file.
404 Not Found: The Endpoint (URL) is wrong or the ID doesn't exist.
429 Too Many Requests: You are spamming the server (Rate Limiting).
500 Internal Server Error: The server crashed. This is not your fault.

```python
import requests 

# 1. The Endpoint
# This is the exact address of the resource we want.
url = "https://jsonplaceholder.typicode.com/users/2"

# 2. The Execution
# Python sends the packet across the internet and waits (blocks) 
# until the server replies.
response = requests.get(url)

# 3. The Response Object

print(f"Server replied with: {response.status_code}")

# 4. Extracting Data
# .text gives us the raw string (useful for HTML)
# .json() parses the string into a Python Dictionary (CRITICAL STEP)
if response.status_code == 200:
    data = response.json()
    
    # Now we use standard Python Dictionary syntax
    print(f"Name: {data['name']}")
    print(f"City: {data['address']['city']}") # Nested access
    print(f"Company: {data['company']['name']}")
else:
    print("Request failed.")
```

HANDLING QUERY PARAMETERS

```python
base_url = "https://jsonplaceholder.typicode.com/posts"

# We want posts by User ID 1 that are NOT completed.
# This mimics: .../posts?userId=1&completed=false
search_criteria = {
    "userId": 1,
    "completed": "false" 
}

# requests constructs the full URL automatically
response = requests.get(base_url, params=search_criteria)

print(response.url) 
# Output: https://jsonplaceholder.typicode.com/posts?userId=1&completed=false
```
